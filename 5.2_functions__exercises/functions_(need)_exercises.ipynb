```python
def calculcate_bmi(weight, height):
    BMI = weight / (height ** 2)
    if BMI < 18.5:
        return 'Underweight'
    if 18.5 <= BMI < 25:
        return 'Normal weight'
    if 25 <= BMI < 30:
        return 'Overweight'
    return 'Obese'
```

```python
def factorial(n):
    if n < 0:
        return 'Error'
    if n == 0 or n == 1:
        return 1
    return n * factorial(n - 1)
```
