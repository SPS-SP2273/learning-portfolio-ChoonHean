```python
def convert_celsius(temperature_celsius, target_scale='Fahrenheit'):
    if target_scale == 'Kelvin':
        return temperature_celsius + 273.15
    return temperature_celsius * 1.8 + 32
```

```python
def convert_fahrenheit(temperature_fahrenheit, target_scale='Celsius'):
    res = (temperature_fahrenheit - 32) / 1.8
    return res + 273.15 if target_scale == 'Kelvin' else res 
```

```python
def convert_kelvin(temperature_kelvin, target_scale='Celsius'):
    res = temperature_kelvin - 273.15
    return res * 1.8 + 32 if target_scale == 'Fahrenheit' else res

def convert_temperature(temperature, source_scale, target_scale):
    if source_scale == 'Celsius':
        return convert_celsius(temperature, target_scale=target_scale)
    elif source_scale == 'Fahrenheit':
        return convert_fahrenheit(temperature, target_scale=target_scale)
    return convert_kelvin(temperature, target_scale=target_scale)
```
