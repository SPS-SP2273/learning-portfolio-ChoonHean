```python
ls = []
for i in range(10):
    if i % 2 == 0:
        continue
    ls.append(i ** 2)
print(ls)
```

```python
%%timeit
ls = []
for i in range(10):
    if i % 2 == 0:
        continue
    ls.append(i ** 2)
```

```python
ls = [i ** 2 for i in range(1, 10, 2)]
print(ls)
```

```python
%%timeit
ls = [i ** 2 for i in range(0, 10, 2)]
```

List comprehension is faster

```python
years = 0
substance = 100
while substance > 50:
    years += 1
    substance = substance * 0.95
print(years)
```
