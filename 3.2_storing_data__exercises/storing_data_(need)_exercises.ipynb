```python
import numpy as np

x_1 = [1, 4, 9, 16, 25, 36, 49, 64, 81, 100]
x_2 = [1, 8, 27, 64, 125, 216, 343, 512, 729, 1000]
u_1 = np.mean(x_1)
u_2 = np.mean(x_2)
sd_1 = np.std(x_1)
sd_2 = np.std(x_2)

r = (sum([(x - u_1) * (y - u_2) for x, y in zip(x_1, x_2)]) / (len(x_1) * sd_1 * sd_2))
     
print(r)
print(np.corrcoef(x_1, x_2))
```

np.corrcoef returns the whole covariance matrix
